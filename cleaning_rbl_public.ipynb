# Step 5: Pre-proessing the data to appropriate data type

```python
df.columns = df.columns.str.strip()
# Converting object to date
df["Batch_Start_Date"]= pd.to_datetime(df["Batch_Start_Date"])
df["Batch_End_Date"]= pd.to_datetime(df["Batch_End_Date"])
df["Placement_Date"]= pd.to_datetime(df["Placement_Date"])
df["If_not_working_last_day_of_job"] = pd.to_datetime(df["If_not_working_last_day_of_job"])
# Converting Integer to String
df["Mobile"]=df["Mobile"].astype(str)
#df["If_yes_Name_of_the_course"]=df["If_yes_Name_of_the_course"].to_string()
df.info()
```
