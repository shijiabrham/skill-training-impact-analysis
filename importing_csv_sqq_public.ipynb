```python
cols = ", ".join([f"`{col}`" for col in df.columns])
placeholders = ", ".join(["%s"] * len(df.columns))

insert_query = f"""
INSERT INTO rbl_data ({cols})
VALUES ({placeholders})
"""

data = df.values.tolist()

cursor.executemany(insert_query, data)

conn.commit()

print("data inserted successfully")
```
